# 9.1 Categorical "None" fill (feature doesn't exist)

````python
none_cols = ['Pool QC', 'Alley', 'Fence', 'Fireplace Qu',
             'Garage Type', 'Garage Finish', 'Garage Qual', 'Garage Cond',
             'Bsmt Qual', 'Bsmt Cond', 'Bsmt Exposure',
             'BsmtFin Type 1', 'BsmtFin Type 2', 'Mas Vnr Type']
for col in none_cols:
    if col in df.columns:
        df[col] = df[col].fillna('None')
print(f"✅ Filled {len(none_cols)} categorical cols with 'None'")
````

<br>
<br>

# 9.2 Numeric 0 fill

````python
zero_cols = ['Mas Vnr Area', 'BsmtFin SF 1', 'BsmtFin SF 2',
             'Bsmt Unf SF', 'Total Bsmt SF', 'Bsmt Full Bath',
             'Bsmt Half Bath', 'Garage Cars', 'Garage Area']
for col in zero_cols:
    if col in df.columns:
        df[col] = df[col].fillna(0)
print(f"✅ Filled {len(zero_cols)} numeric cols with 0")
````

<br>
<br>

# 9.3 Garage Yr Blt → Year Built when no garage

````python
df['Garage Yr Blt'] = df['Garage Yr Blt'].fillna(df['Year Built'])
print("✅ Filled Garage Yr Blt with Year Built")
````

<br>

````python
df['Lot Frontage'] = df.groupby('Neighborhood')['Lot Frontage']\
                       .transform(lambda x: x.fillna(x.median()))
print("✅ Filled Lot Frontage by Neighborhood median")
````

# 9.5 Remaining categorical → mode

````python
for col in df.select_dtypes(include='object').columns:
    if df[col].isnull().sum() > 0:
        df[col] = df[col].fillna(df[col].mode()[0])
print("✅ Filled remaining categorical with mode")
````

<br>
<br>

# 9.6 Remaining numeric → median

````python
for col in df.select_dtypes(include=np.number).columns:
    if df[col].isnull().sum() > 0:
        df[col] = df[col].fillna(df[col].median())
print("✅ Filled remaining numeric with median")

print(f"\n🔍 Total missing after handling: {df.isnull().sum().sum()}")
````

# **11.1: IQR Detection**

````python
# ---------- 9.1 IQR Detection ----------
def detect_outliers_iqr(data, col):
    Q1, Q3 = data[col].quantile([0.25, 0.75])
    IQR = Q3 - Q1
    lower, upper = Q1 - 1.5 * IQR, Q3 + 1.5 * IQR
    return data[(data[col] < lower) | (data[col] > upper)], lower, upper

print("\n--- 9.1 IQR Outlier Detection ---")
outlier_summary = []
for col in ['Gr Liv Area', 'Lot Area', 'SalePrice', 'Total Bsmt SF']:
    outliers, lo, hi = detect_outliers_iqr(df, col)
    outlier_summary.append({
        'Feature': col,
        'Outliers': len(outliers),
        'Lower Bound': round(lo, 0),
        'Upper Bound': round(hi, 0),
        'Outlier %': round(len(outliers) / len(df) * 100, 2)
    })
outlier_df = pd.DataFrame(outlier_summary)
outlier_df
````

♠ Step-15: Feature Scaling🏘️</h3>

````python
# ============================================================
# STEP 13: FEATURE SCALING (MODEL-AWARE) — ENHANCED
# ============================================================
print("\n" + "=" * 60)
print("STEP 13: FEATURE SCALING (MODEL-AWARE)")
print("=" * 60)

from sklearn.preprocessing import StandardScaler

# ------------------------------------------------------------
# 13.1 Fit Scaler on TRAINING data only (avoid leakage)
# ------------------------------------------------------------
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print("✅ StandardScaler fitted on training data only (no leakage)")
print(f"   Train shape: {X_train_scaled.shape}")
print(f"   Test shape : {X_test_scaled.shape}")

# ------------------------------------------------------------
# 13.2 Scaling Statistics (verify)
# ------------------------------------------------------------
print("\n--- Scaling Statistics (Sample Features) ---")
sample_cols = ['Lot Area', 'Gr Liv Area', 'Overall Qual', 'Year Built']
sample_idx = [X_train.columns.get_loc(c) for c in sample_cols 
              if c in X_train.columns]

stats_df = pd.DataFrame({
    'Feature': [X_train.columns[i] for i in sample_idx],
    'Original Mean': [X_train.iloc[:, i].mean() for i in sample_idx],
    'Original Std': [X_train.iloc[:, i].std() for i in sample_idx],
    'Scaled Mean': [X_train_scaled[:, i].mean() for i in sample_idx],
    'Scaled Std': [X_train_scaled[:, i].std() for i in sample_idx]
})
print(stats_df.round(4).to_string(index=False))

# ------------------------------------------------------------
# 13.3 Visualize Scaling Effect
# ------------------------------------------------------------
fig, axes = plt.subplots(2, 2, figsize=(14, 10))
for ax, col in zip(axes.flatten(), sample_cols[:4]):
    if col in X_train.columns:
        idx = X_train.columns.get_loc(col)
        ax.hist(X_train.iloc[:, idx], bins=30, alpha=0.7, 
                color='salmon', label='Original')
        ax.hist(X_train_scaled[:, idx], bins=30, alpha=0.7, 
                color='steelblue', label='Scaled')
        ax.set_title(f'{col}: Before vs After Scaling')
        ax.legend()
        ax.set_xlabel('Value')
        ax.set_ylabel('Frequency')
plt.suptitle('Effect of StandardScaler on Feature Distributions', 
             fontsize=14, y=1.02)
plt.tight_layout()
plt.show()

# ------------------------------------------------------------
# 13.4 Model-Aware Scaling Strategy
# ------------------------------------------------------------
print("\n--- Model-Aware Scaling Strategy ---")

# Tree-based models: scale-invariant
tree_based_models = [
    'Decision Tree',
    'Random Forest',
    'Extra Trees',
    'Gradient Boosting',
    'XGBoost',
    'LightGBM',
    'CatBoost'
]

# Scale-sensitive models
scale_sensitive_models = [
    'Linear Regression',
    'Ridge',
    'Lasso',
    'ElasticNet',
    'KNN',
    'SVR',
    'Neural Network',
    'PCA + Linear'
]

print("\n✅ Tree-based models (use UNSCALED data):")
for m in tree_based_models:
    print(f"   • {m}")

print("\n✅ Scale-sensitive models (use SCALED data):")
for m in scale_sensitive_models:
    print(f"   • {m}")

# ------------------------------------------------------------
# 13.5 Justification
# ------------------------------------------------------------
print("\n--- 13.5 Scaling Justification ---")
print("""
📌 Why Feature Scaling is needed for Ames Dataset:

Dataset Magnitude Range:
   • Lot Area          : 1,300 – 215,245   (10^5)
   • Total SF          : 334 – 6,110       (10^3)
   • Year Built        : 1872 – 2010       (10^3)
   • Overall Qual      : 1 – 10            (10^0)
   • Full Bath         : 0 – 4             (10^0)

⚠️ Without scaling:
   • Linear Regression: biased toward high-magnitude features
   • Ridge/Lasso: regularization penalty unfair across features
   • KNN: Lot Area dominates distance metric
   • SVR: RBF kernel sensitive to magnitude

✅ With scaling (StandardScaler):
   • All features → mean = 0, std = 1
   • Fair contribution in gradient descent
   • Distance metric balanced
   • Regularization balanced

❌ Not needed for:
   • Tree-based models (split on thresholds, not distances)
   • Example: Random Forest, XGBoost — scaling gives identical result
""")
````

♠ Step-16: ML Model Training🏘️</h3>

````python
# ============================================================
# STEP 14: ML MODEL TRAINING (MODEL-AWARE)
# ============================================================
print("\n" + "=" * 60)
print("STEP 14: ML MODEL TRAINING (MODEL-AWARE)")
print("=" * 60)

from sklearn.linear_model import LinearRegression, Ridge, Lasso
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor

try:
    from xgboost import XGBRegressor
    xgb_available = True
except ImportError:
    xgb_available = False
    print("⚠️  XGBoost not installed. Skipping.")

# ------------------------------------------------------------
# Define models with their scaling requirement
# ------------------------------------------------------------
models_config = {
    # Linear models — need SCALED data
    'Linear Regression': {
        'model': LinearRegression(),
        'scaled': True
    },
    'Ridge': {
        'model': Ridge(alpha=1.0, random_state=40),
        'scaled': True
    },
    'Lasso': {
        'model': Lasso(alpha=0.001, random_state=42),
        'scaled': True
    },
    # Tree models — use UNSCALED data
    'Random Forest': {
        'model': RandomForestRegressor(n_estimators=200, 
                                       random_state=42, 
                                       n_jobs=-1),
        'scaled': False
    },
    'Gradient Boosting': {
        'model': GradientBoostingRegressor(random_state=42),
        'scaled': False
    },
}

if xgb_available:
    models_config['XGBoost'] = {
        'model': XGBRegressor(n_estimators=200, 
                              random_state=42, 
                              n_jobs=-1),
        'scaled': False
    }

# ------------------------------------------------------------
# Train each model with appropriate data
# ------------------------------------------------------------
print("\n--- Training Models ---")
trained_models = {}

for name, config in models_config.items():
    model = config['model']
    
    # Pick scaled or unscaled based on model type
    if config['scaled']:
        X_tr = X_train_scaled
        data_type = "SCALED"
    else:
        X_tr = X_train
        data_type = "UNSCALED"
    
    model.fit(X_tr, y_train)
    trained_models[name] = config  # save model + scaling flag
    print(f"✅ {name:22s} trained on {data_type:8s} data")
````

♠ Step-17: Model Evaluation🏘️</h3>

````python
# ============================================================
# STEP 15: MODEL EVALUATION (MODEL-AWARE)
# ============================================================
print("\n" + "=" * 60)
print("STEP 15: MODEL EVALUATION")
print("=" * 60)

from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

results = []
for name, config in trained_models.items():
    model = config['model']
    
    # Use appropriate test data based on model type
    if config['scaled']:
        X_te = X_test_scaled
    else:
        X_te = X_test
    
    pred = model.predict(X_te)
    mae = mean_absolute_error(y_test, pred)
    rmse = np.sqrt(mean_squared_error(y_test, pred))
    r2 = r2_score(y_test, pred)
    
    results.append({
        'Model': name,
        'Scaling': 'Scaled' if config['scaled'] else 'Unscaled',
        'MAE': round(mae, 2),
        'RMSE': round(rmse, 2),
        'R²': round(r2, 4)
    })

results_df = pd.DataFrame(results).sort_values('RMSE')
print("\n--- Model Comparison (Model-Aware Scaling) ---")
print(results_df.to_string(index=False))

best_name = results_df.iloc[0]['Model']
print(f"\n🏆 Best Model: {best_name}")

# Visualization
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
results_df.plot(x='Model', y='RMSE', kind='barh', ax=axes[0],
                color='coral', legend=False)
axes[0].set_title('RMSE by Model')
results_df.plot(x='Model', y='R²', kind='barh', ax=axes[1],
                color='seagreen', legend=False)
axes[1].set_title('R² by Model')
plt.tight_layout()
plt.show()
````

````python
# ============================================================
# STEP 16: HYPERPARAMETER TUNING
# ============================================================
print("\n" + "=" * 60)
print("STEP 16: HYPERPARAMETER TUNING")
print("=" * 60)

from sklearn.model_selection import GridSearchCV

# Determine which data to use (based on best model type)
best_config = trained_models[best_name]
X_tr_final = X_train_scaled if best_config['scaled'] else X_train
X_te_final = X_test_scaled  if best_config['scaled'] else X_test
print(f"Using {'SCALED' if best_config['scaled'] else 'UNSCALED'} data for tuning")

# Define param grid based on best model
if best_name == 'Random Forest':
    param_grid = {
        'n_estimators': [100, 200, 300],
        'max_depth': [10, 20, None],
        'min_samples_split': [2, 5, 10]
    }
    base_model = RandomForestRegressor(random_state=42, n_jobs=-1)
    
elif best_name == 'Gradient Boosting':
    param_grid = {
        'n_estimators': [100, 200, 300],
        'max_depth': [3, 5, 7],
        'learning_rate': [0.05, 0.1, 0.2]
    }
    base_model = GradientBoostingRegressor(random_state=42)
    
elif best_name == 'XGBoost':
    param_grid = {
        'n_estimators': [100, 200, 300],
        'max_depth': [3, 5, 7],
        'learning_rate': [0.05, 0.1, 0.2]
    }
    base_model = XGBRegressor(random_state=42, n_jobs=-1)
    
else:
    param_grid = {}
    base_model = best_config['model']

# Run GridSearch
if param_grid:
    grid = GridSearchCV(
        base_model, param_grid, cv=5,
        scoring='neg_mean_squared_error', n_jobs=-1, verbose=1
    )
    grid.fit(X_tr_final, y_train)
    print(f"\n✅ Best Params: {grid.best_params_}")
    best_model = grid.best_estimator_
else:
    best_model = base_model

# Final evaluation
final_pred = best_model.predict(X_te_final)
print("\n--- Final Model Performance ---")
print(f"MAE : {mean_absolute_error(y_test, final_pred):.4f}")
print(f"RMSE: {np.sqrt(mean_squared_error(y_test, final_pred)):.4f}")
print(f"R²  : {r2_score(y_test, final_pred):.4f}")
````
